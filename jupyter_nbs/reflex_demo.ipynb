# Reflex demo: requests module
This notebook runs the updated Reflex extractors against the `requests` package using mypy stubgen for signatures/types, then previews and visualizes the results.

```python
# Environment & imports
import os, sys, time
# Add repo root so we can import backend.code_sage.reflex
repo_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if repo_root not in sys.path:
    sys.path.insert(0, repo_root)

from backend.code_sage.reflex import process_module, process_package, visualize_to_html
import importlib
import importlib.metadata as im

# Ensure 'requests' can be imported (should already be available per requirements)
try:
    import requests  # noqa
except Exception as e:
    raise RuntimeError(f'Failed to import requests: {e}')

# Resolve version for labeling
try:
    requests_version = im.version('requests')
except Exception:
    # Fallback to attribute if metadata is not available
    import requests as _rq
    requests_version = getattr(_rq, '__version__', None) or ''

requests_version
```

```python
# Run extraction
start = time.time()
entities, relations, other = process_module('pydantic', version=requests_version)
elapsed = time.time() - start
print(f'Extraction completed in {elapsed:.2f}s')
print(f'Entities: {len(entities)} | Relations: {len(relations)}')
# Show other-relations summary sizes
for k, v in other.items():
    try:
        print(f"{k}: {len(v)}")
    except TypeError:
        print(f"{k}: {v}")
```

```python
# Search for entity given name
search_name = 'pydantic.fields.Field'
found = [e for e in entities if e.name.startswith(search_name)]
print(f'Found {len(found)} entities matching "{search_name}":')
for e in found:
    print(e.name)
    print('  labels:', e.labels)
    if 'signature' in e.properties:
        print('  signature:', e.properties.get('signature'))
    if 'resolved_signature' in e.properties:
        print('  resolved_signature:', e.properties.get('resolved_signature'))
    if 'overload_signatures' in e.properties:
        print('  overload_signatures:', e.properties.get('overload_signatures'))
    if 'resolved_overload_signatures' in e.properties:
        print('  resolved_overload_signatures:', e.properties.get('resolved_overload_signatures'))
    if 'source' in e.properties:
        print('  source:', e.properties.get('source'))
    if 'found_types' in e.properties:
        print('  found_types:', e.properties.get('found_types'))
    print()
```

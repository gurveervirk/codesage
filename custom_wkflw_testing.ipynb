```python
import ast
from pathlib import Path
from collections import defaultdict

class ScopedTypeAnalyzer(ast.NodeVisitor):
    def __init__(self):
        self.imports = {}  # alias → full module path
        self.defined_funcs = set()
        self.defined_classes = {}
        self.external_calls = []
        self.internal_calls = []

        self.scope_stack = []  # function/class names
        self.scope_vars = [{}]  # stack of var_name → [types]

        self.var_usages = defaultdict(list)  # (scope, var) → list of usages

        self.class_stack = []
        self.func_param_types = {}  # func name → {param: type}

        self.scoped_var_usages = defaultdict(lambda: defaultdict(list))  # func → var → [usages]

        # Remove resolved_calls, add call_var_map for mapping variables in calls
        self.call_var_map = []

    def current_scope(self):
        return ".".join(self.scope_stack) if self.scope_stack else "<module>"

    def current_vars(self):
        return self.scope_vars[-1]

    def visit_Import(self, node):
        for alias in node.names:
            self.imports[alias.asname or alias.name] = alias.name
        self.generic_visit(node)

    def visit_ImportFrom(self, node):
        module = node.module or ''
        for alias in node.names:
            full_name = f"{module}.{alias.name}"
            self.imports[alias.asname or alias.name] = full_name
        self.generic_visit(node)

    def visit_ClassDef(self, node):
        self.scope_stack.append(node.name)
        self.scope_vars.append({})
        parent_names = []

        for base in node.bases:
            base_name = self.get_full_name(base)
            if base_name:
                parent_names.append(base_name)

        self.defined_classes[node.name] = parent_names
        self.generic_visit(node)

        self.scope_vars.pop()
        self.scope_stack.pop()

    def visit_FunctionDef(self, node):
        self.scope_stack.append(node.name)
        self.scope_vars.append({})

        # Parameter types
        param_types = {}
        for arg in node.args.args:
            param_line = getattr(arg, "lineno", node.lineno)
            if arg.annotation:
                type_str = self.get_full_name(arg.annotation)
                if type_str:
                    param_types[arg.arg] = type_str
                    self.current_vars()[arg.arg] = [type_str]  # seed var scope
                    # Add parameter usage at definition
                    self.scoped_var_usages[self.current_scope()][arg.arg].append({
                        'line': param_line,
                        'type': type_str,
                        'usage': 'definition'
                    })
                else:
                    # Add parameter usage at definition even if type unknown
                    self.scoped_var_usages[self.current_scope()][arg.arg].append({
                        'line': param_line,
                        'type': None,
                        'usage': 'definition'
                    })
            else:
                # Add parameter usage at definition even if type unknown
                self.scoped_var_usages[self.current_scope()][arg.arg].append({
                    'line': param_line,
                    'type': None,
                    'usage': 'definition'
                })

        self.func_param_types[node.name] = param_types

        self.defined_funcs.add(node.name)

        self.generic_visit(node)

        self.scope_vars.pop()
        self.scope_stack.pop()

    def visit_Assign(self, node):
        if isinstance(node.value, ast.Call):
            type_name = self.get_full_name(node.value.func)
            if type_name:
                for target in node.targets:
                    var_name = self.get_var_name(target)
                    if var_name:
                        current_types = self.current_vars().setdefault(var_name, [])
                        if type_name not in current_types:
                            current_types.append(type_name)
        self.generic_visit(node)

    def visit_Call(self, node):
        full_func = self.get_full_name(node.func)
        scope = self.current_scope()

        # Instead of resolving, map variables in calls to their definitions/usages
        call_vars = []
        for arg in node.args:
            if isinstance(arg, ast.Name):
                call_vars.append(arg.id)
            elif isinstance(arg, ast.Attribute):
                # e.g., obj.method()
                if isinstance(arg.value, ast.Name):
                    call_vars.append(arg.value.id)
        # For keyword arguments
        for kw in node.keywords:
            if isinstance(kw.value, ast.Name):
                call_vars.append(kw.value.id)
            elif isinstance(kw.value, ast.Attribute):
                if isinstance(kw.value.value, ast.Name):
                    call_vars.append(kw.value.value.id)

        # Store mapping: call line, call name, variables, and their usages if available
        var_mappings = []
        for var in call_vars:
            usages = self.var_usages.get((scope, var), [])
            var_mappings.append({
                'var': var,
                'usages': usages
            })
        self.call_var_map.append({
            'line': node.lineno,
            'call': full_func,
            'vars': var_mappings
        })

        # Categorize as external/internal
        if full_func:
            top = full_func.split('.')[0]
            call_info = {
                'line': node.lineno,
                'call': full_func,
                'args': len(node.args),
                'keywords': [kw.arg for kw in node.keywords if kw.arg]
            }
            if top in self.defined_funcs or top in self.defined_classes:
                self.internal_calls.append(call_info)
            else:
                self.external_calls.append(call_info)

        self.generic_visit(node)

    def get_var_name(self, node):
        if isinstance(node, ast.Name):
            return node.id
        elif isinstance(node, ast.Attribute):
            if isinstance(node.value, ast.Name) and node.value.id == 'self':
                return f"{self.class_stack[-1]}.{node.attr}" if self.class_stack else f"self.{node.attr}"
        return None

    def lookup_var_types(self, var_name):
        for scope_vars in reversed(self.scope_vars):
            if var_name in scope_vars:
                return scope_vars[var_name]
        return []

    def get_full_name(self, node):
        if isinstance(node, ast.Name):
            return self.imports.get(node.id, node.id)
        elif isinstance(node, ast.Attribute):
            parts = []
            while isinstance(node, ast.Attribute):
                parts.append(node.attr)
                node = node.value
            if isinstance(node, ast.Name):
                base = self.imports.get(node.id, node.id)
                parts.append(base)
                return ".".join(reversed(parts))
        elif isinstance(node, ast.Subscript):
            return self.get_full_name(node.value)
        elif isinstance(node, ast.Call):
            return self.get_full_name(node.func)
        return None

def analyze_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        source = f.read()

    tree = ast.parse(source, filename=file_path)
    analyzer = ScopedTypeAnalyzer()
    analyzer.visit(tree)

    # Output call_var_map instead of resolved_calls
    return {
        'file': file_path,
        'external_calls': analyzer.external_calls,
        'internal_calls': analyzer.internal_calls,
        'defined_classes': analyzer.defined_classes,
        'defined_functions': list(analyzer.defined_funcs),
        'param_types': analyzer.func_param_types,
        'scoped_var_usages': {k: dict(v) for k, v in analyzer.scoped_var_usages.items()},
        'call_var_map': analyzer.call_var_map,
    }
```

```python
import ast
import json
from collections import defaultdict

BUILTINS = set(dir(__builtins__))

class Scope:
    """Represents a scope, holding rich information about variables."""
    def __init__(self, name: str, parent=None):
        self.name = name
        self.parent = parent
        self.variables = {} # Maps var_name -> {'type': str, 'source_str': str, 'lineno': int}

    def __repr__(self):
        return f"<Scope: {self.name}, Vars: {list(self.variables.keys())}>"

class FullyResolvingAnalyzer(ast.NodeVisitor):
    """
    Analyzes code to produce a detailed report where all aliases and variables
    are substituted with their fully-resolved canonical paths.
    """
    def __init__(self):
        self.call_report = defaultdict(list)
        self.variable_report = defaultdict(dict)
        self.scope_stack = []
        self.defined_in_file = set()
        self.import_map = {}
        self._push_scope('global')

    # --- Scope Management ---
    def _push_scope(self, name: str):
        parent = self.scope_stack[-1] if self.scope_stack else None
        self.scope_stack.append(Scope(name, parent))

    def _pop_scope(self):
        if self.scope_stack:
            popped_scope = self.scope_stack.pop()
            self.variable_report[popped_scope.name] = popped_scope.variables
            return popped_scope

    @property
    def current_scope(self):
        return self.scope_stack[-1]

    # --- Helper Functions ---
    def _node_to_str(self, node):
        if node is None: return "None"
        try: return ast.unparse(node)
        except AttributeError:
            if isinstance(node, ast.Name): return node.id
            if isinstance(node, ast.Constant): return repr(node.value)
            return f"<{type(node).__name__}>"

    def _find_variable_info(self, name: str):
        scope = self.current_scope
        while scope:
            if name in scope.variables: return scope.variables[name]
            scope = scope.parent
        return None

    def _get_canonical_path(self, node):
        """Returns the single, fully-resolved canonical string path for any expression."""
        if isinstance(node, ast.Name):
            var_info = self._find_variable_info(node.id)
            if var_info: return var_info['type']
            if node.id in self.import_map: return self.import_map[node.id]
            return node.id
        if isinstance(node, ast.Attribute):
            base_type = self._get_canonical_path(node.value)
            return f"{base_type}.{node.attr}"
        if isinstance(node, ast.Call):
            return self._get_canonical_path(node.func)
        return self._node_to_str(node)

    def _store_variable(self, var_name, type_node, value_node, lineno):
        inferred_type = self._get_canonical_path(type_node or value_node)
        source_str = self._node_to_str(value_node) if value_node else None
        self.current_scope.variables[var_name] = {
            'type': inferred_type, 'source_str': source_str, 'lineno': lineno
        }

    def _resolve_and_structure_node(self, node):
        """
        Recursively builds a structured dictionary representing a node,
        fully resolving all names to their canonical paths.
        """
        if isinstance(node, ast.Constant):
            return {'param_type': 'literal', 'value': repr(node.value)}
        
        if isinstance(node, ast.Name):
            var_info = self._find_variable_info(node.id)
            if var_info:
                return {'param_type': 'resolved_variable', 'name': node.id, 'resolved_type': var_info['type']}
            if node.id in self.import_map:
                return {'param_type': 'resolved_alias', 'name': node.id, 'full_path': self.import_map[node.id]}
            return {'param_type': 'unresolved_name', 'name': node.id}

        if isinstance(node, ast.Attribute):
            return {'param_type': 'attribute_access', 'base': self._resolve_and_structure_node(node.value), 'attr': node.attr}
        
        if isinstance(node, ast.Call):
            return {'param_type': 'call', 'function': self._resolve_and_structure_node(node.func), 'args': [self._resolve_and_structure_node(a) for a in node.args], 'kwargs': {kw.arg: self._resolve_and_structure_node(kw.value) for kw in node.keywords}}

        if isinstance(node, (ast.List, ast.Tuple)):
            return {'param_type': 'list' if isinstance(node, ast.List) else 'tuple', 'elements': [self._resolve_and_structure_node(e) for e in node.elts]}
            
        if isinstance(node, ast.Dict):
            return {'param_type': 'dict', 'items': {self._node_to_str(k): self._resolve_and_structure_node(v) for k, v in zip(node.keys, node.values)}}

        return {'param_type': 'expression', 'value': self._node_to_str(node)}

    # --- Visitor Methods ---
    def visit_Import(self, node: ast.Import):
        for alias in node.names: self.import_map[alias.asname or alias.name] = alias.name

    def visit_ImportFrom(self, node: ast.ImportFrom):
        module_path = node.module or ""
        for alias in node.names: self.import_map[alias.asname or alias.name] = f"{module_path}.{alias.name}"

    def visit_FunctionDef(self, node: ast.FunctionDef):
        self.defined_in_file.add(node.name)
        self.current_scope.variables[node.name] = {'type': 'function', 'source_str': f'def {node.name}(...)', 'lineno': node.lineno}
        self._push_scope(f"function:{node.name}")
        if node.args.args and node.args.args[0].arg in ('self', 'cls'):
            class_scope = next((s for s in reversed(self.scope_stack[:-1]) if s.name.startswith('class:')), None)
            if class_scope: self._store_variable(node.args.args[0].arg, ast.Name(id=class_scope.name.split(':')[1]), None, node.lineno)
        all_args = []
        if hasattr(node.args, 'posonlyargs'): all_args.extend(node.args.posonlyargs)
        if hasattr(node.args, 'args'): all_args.extend(node.args.args)
        if hasattr(node.args, 'kwonlyargs'): all_args.extend(node.args.kwonlyargs)
        for arg in all_args: self._store_variable(arg.arg, arg.annotation, None, node.lineno)
        self.generic_visit(node)
        self._pop_scope()

    def visit_ClassDef(self, node: ast.ClassDef):
        self.defined_in_file.add(node.name)
        self.current_scope.variables[node.name] = {'type': 'class', 'source_str': f'class {node.name}(...)', 'lineno': node.lineno}
        self._push_scope(f"class:{node.name}")
        self.generic_visit(node)
        self._pop_scope()

    def visit_Assign(self, node: ast.Assign):
        for target in node.targets:
            if isinstance(target, ast.Name): self._store_variable(target.id, None, node.value, node.lineno)
        self.generic_visit(node)

    def visit_AnnAssign(self, node: ast.AnnAssign):
        if isinstance(node.target, ast.Name): self._store_variable(node.target.id, node.annotation, node.value, node.lineno)
        self.generic_visit(node)

    def visit_Call(self, node: ast.Call):
        full_call_path = self._get_canonical_path(node.func)
        root_type = full_call_path.split('.')[0]
        call_type = "internal" if root_type in self.defined_in_file else "external"
        if '.' not in full_call_path and full_call_path in BUILTINS: call_type = "external"
            
        params = []
        for i, arg in enumerate(node.args): params.append({"name": f"arg_{i}", "arg_info": self._resolve_and_structure_node(arg)})
        for kw in node.keywords: params.append({"name": kw.arg, "arg_info": self._resolve_and_structure_node(kw.value)})

        self.call_report[self.current_scope.name].append({"call": full_call_path, "lineno": node.lineno, "type": call_type, "params": params})
        self.generic_visit(node)

def run_analysis(source_code: str) -> dict:
    try:
        tree = ast.parse(source_code)
        analyzer = FullyResolvingAnalyzer()
        analyzer.visit(tree)
        analyzer._pop_scope()
        return {'variables': dict(analyzer.variable_report), 'calls': dict(analyzer.call_report)}
    except Exception as e:
        print(f"An error occurred during analysis: {type(e).__name__}: {e}")
        return {}

# --- Example Usage ---
if __name__ == "__main__":
    sample_code = """
import aerosandbox as asb
import aerosandbox.numpy as np

# Define the geometry of the biplane
wing = asb.Wing(
    symmetric=True,
    xsecs=[
        asb.Airfoil("naca4412").translate(xyz_nondim=[0, 0, 0]),
        asb.Airfoil("naca4412").translate(xyz_nondim=[0.02, 0.5, 0]) # Add a slight stagger
    ],
    chord=1,
    span=10
)

horizontal_stabilizer = asb.Wing(
    symmetric=True,
    xsecs=[
        asb.Airfoil("naca0012").translate(xyz_nondim=[0, 0, 0]),
        asb.Airfoil("naca0012").translate(xyz_nondim=[0, 0.5, 0])  #No stagger
    ],
    chord=0.7,
    span=8
)

airplane = asb.Airplane(
    wings=[wing, horizontal_stabilizer.translate(xyz=[4, 0, -2])]
)

# Define the operating point
op_point = asb.OperatingPoint(
    velocity=30,  # m/s
    alpha=5,  # degrees
    beta=0,  # degrees
    atmosphere=asb.Atmosphere(altitude=0) # Sea level
)

# Run the Vortex Lattice Method analysis
analysis = asb.VortexLatticeMethod(
    airplane=airplane,
    op_point=op_point,
    spanwise_resolution=16, # Increase for accuracy
    chordwise_resolution=8  # Increase for accuracy
)

aero_data = analysis.run()

# Print the results
print(aero_data)

# Access the drag coefficient
CD = aero_data["CD"]
print(f"Drag Coefficient: {CD}")
"""
    
    analysis_result = run_analysis(sample_code)
    if analysis_result:
        print("--- VARIABLE MAP ---")
        print(json.dumps(analysis_result.get('variables'), indent=2))
        print("\n\n--- CALL REPORT ---")
        print(json.dumps(analysis_result.get('calls'), indent=2))
```
